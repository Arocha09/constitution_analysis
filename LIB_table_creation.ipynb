```python
import pandas as pd
import numpy as np
import os
import json
```

```python
with open('constitutions.json', 'r') as file:
    constitutions = json.load(file)

dem_data = pd.read_csv('data/merged_democracy_data.csv')
```

## CREATION OF LIB

```python
lib_list = []
for key in constitutions.keys():
    name_part, date_part = key.rsplit('_', 1)
    country_name = name_part.replace('_', ' ')
    lib_list.append({
    'doc_id': key,
    'country_name': country_name,
    'doc_year': int(date_part),
    'doc_len': len(constitutions[key])
    })

LIB = pd.DataFrame(lib_list)



# 2. Merge LIB with only the latest democracy stats
merged_df = pd.merge(LIB, dem_data, on='country_name', how='left')

```

```python
merged_df.head()
```

```python
merged_df.set_index('doc_id', inplace=True)
```

```python
LIB = merged_df
```

```python
LIB = LIB.rename({'doc_year' : 'year_amended', 'year' : 'dem_year'}, axis=1)
```
