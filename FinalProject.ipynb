- Description of TFIDIF formula ($\LaTeX$ OK):

TF-IDF values were computed using `sklearn.feature_extraction.text.TfidfTransformer` with `norm=None`.

For term \( $t$ \) in document \( $d$ \):

$\text{tfidf}(t,d) = \text{tf}(t,d) \times \text{idf}(t)$

where

$\text{idf}(t) = \log\left(\frac{1 + N}{1 + df(t)}\right) + 1$

- \( $\text{tf}(t,d)$ \) = raw frequency of term \( $t$ \) in document \( $d$ \)
- \( $N$ \) = total number of documents
- \( $df(t)$ \) = number of documents containing term \( $t$ \)

No vector normalization was applied.

Used parquet once again because it is a large table.
