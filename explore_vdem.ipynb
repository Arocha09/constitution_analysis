```python
regime_map = {
    0: "Closed autocracy",
    1: "Electoral autocracy",
    2: "Electoral democracy",
    3: "Liberal democracy"
}

vdem_2024['regime_type'] = vdem_2024['v2x_regime'].map(regime_map)
```

```python
vdem_2024
```

```python
dem_table = vdem_2024[['country_name', 'country_text_id', 'year', 'regime_type']]
```

```python
dem_index.head()
```

```python
dem_index_2024 = dem_index[dem_index['Year']==2024]
```

```python
dem_index_2024
```

```python
merged = pd.merge(
    dem_table,
    dem_index_2024,
    left_on='country_name',
    right_on='Entity',
    how='left'
)
```

```python
merged.head()
```

```python
merged = merged.drop(['Code', 'Year', 'Entity'], axis=1)
```
